## Download data from gdrive

```python
import gdown
```

```python
url = "https://drive.google.com/file/d/1layyK2U4Z2XaP-4YsKd52f40U1kqdKaS/view?usp=sharing"
```

```python
file_id = url.split('/')[5]
```

```python
prefix = "https://drive.google.com//uc?export=download&id="
gdown.download(prefix+file_id, "kidney-ct-scan.zip")
```
